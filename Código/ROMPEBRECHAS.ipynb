### <font color=green> 8. CREAR UNA TABLA RESUMEN SIMPLE PARA CADA FUENTE </font>

```python
def resumen_fuente(datos, nombre_fuente):
    """Tabla resumen por variable. Un valor se cuenta como faltante si es NaN
    o si es una cadena vacia (que es como el formato DBF codifica el vacio)."""
    vacio = datos.isna() | (datos.astype(str).apply(lambda s: s.str.strip()) == "")
    return pd.DataFrame({
        "fuente": nombre_fuente,
        "variable": datos.columns,
        "tipo_python": datos.dtypes.astype(str).values,
        "n_faltantes": vacio.sum().values,
        "porcentaje_faltantes": (vacio.mean() * 100).round(2).values,
        "valores_distintos": datos.nunique(dropna=True).values,
    })

resumen_f1 = resumen_fuente(df,  "F1_Padron_web")
resumen_f2 = resumen_fuente(df1, "F2_Lineal_3AP_1")
resumen_f3 = resumen_fuente(df2, "DenunciasEscolares")

tabla_resumen_fuentes = pd.concat([resumen_f1, resumen_f2, resumen_f3], ignore_index=True)

print("F1 - Padron_web (45 variables)")
display(resumen_f1)
print("F2 - Lineal_3AP_1 (230 variables) - primeras 30")
display(resumen_f2.head(30))
print("denuncias_escolares.xlsx ( 8 variables)")
display(resumen_f3)
print("Tabla resumen combinada:", tabla_resumen_fuentes.shape)
```

### <font color=green> 10. INTEGRAR LAS DOS FUENTES </font>

```python
# Llave de integracion: COD_MOD + ANEXO (NO COD_MOD solo: hay 1,409 codigos
# repetidos en F1 y ANEXO es constante '0' en F2, asi que unir por COD_MOD
# perderia o duplicaria silenciosamente los servicios con anexo distinto de 0).

base_integrada = pd.merge(
    df,
    df1,
    on=["COD_MOD", "ANEXO"],
    how="outer",
    suffixes=("_f1", "_f2"),
    indicator=True,
)

print("Base integrada:", base_integrada.shape)
display(base_integrada["_merge"].value_counts())

# Version util para el analisis: solo Inicial escolarizada activa, que es
# donde F1 y F2 se solapan de verdad.
inicial = base_integrada[
    (base_integrada["D_ESTADO"] == "Activo")
    & (base_integrada["NIV_MOD_f1"].isin(["A1", "A2", "A3"]))
].copy()

print("\nInicial escolarizada activa:", len(inicial))
print("...con datos de la Cedula 1A: %d (%.1f%% de cobertura del censo)"
      % ((inicial["_merge"] == "both").sum(),
         (inicial["_merge"] == "both").mean() * 100))

display(base_integrada.head())
```

```python
# ===========================================================================
# PLANTILLA DE DICCIONARIO DE DATOS
# F1 (45 columnas) se documenta una por una a mano.
# F2 (230 columnas) se documenta por bloques y por patron de nomenclatura,
#     porque son las preguntas de la Cedula 1A del Censo Escolar.
# ===========================================================================
import re

# --- 1. F1: descripciones curadas -----------------------------------------
# formato: variable -> (descripcion, tipo_general, escala, rol_o_subtipo, observaciones)
DESC_F1 = {
 "CODINST":   ("Codigo de la institucion educativa (agrupa varios servicios)", "Categorica", "Nominal", "Identificador", "47.14% vacio; no usar como llave"),
 "COD_MOD":   ("Codigo modular del servicio educativo", "Categorica", "Nominal", "Identificador", "7 digitos con ceros a la izquierda; NO es unico (1,409 repetidos)"),
 "ANEXO":     ("Numero de anexo del servicio", "Categorica", "Nominal", "Identificador", "Completa a COD_MOD para formar la llave primaria real"),
 "CODLOCAL":  ("Codigo del local escolar donde funciona el servicio", "Categorica", "Nominal", "Identificador", "21.96% vacio; 69.78% vacio en no escolarizada -> sesgo"),
 "CEN_EDU":   ("Nombre del servicio educativo", "Categorica", "Nominal", "Texto", "Texto libre, 81,916 valores distintos"),
 "NIV_MOD":   ("Codigo de nivel y modalidad educativa", "Categorica", "Nominal", "Codigo", "29 codigos. A1 Cuna, A2 Jardin, A3 Cuna-jardin, A5 Inicial no escolarizado, B0 Primaria, F0 Secundaria"),
 "D_NIV_MOD": ("Descripcion del nivel y modalidad", "Categorica", "Nominal", "Descripcion de codigo", "21 descripciones para 29 codigos (L1-L9 colapsan en una sola)"),
 "D_FORMA":   ("Forma de atencion del servicio", "Categorica", "Nominal", "Descripcion de codigo", "Escolarizada / No escolarizada / No aplica"),
 "TIPSSEXO":  ("Codigo de sexo de los estudiantes atendidos", "Categorica", "Nominal", "Codigo", "1 Varones, 2 Mujeres, 3 Mixto"),
 "D_TIPSSEXO":("Descripcion del sexo atendido", "Categorica", "Nominal", "Descripcion de codigo", "99.0% Mixto: baja varianza"),
 "GESTION":   ("Codigo de tipo de gestion", "Categorica", "Nominal", "Codigo", "1 Publica gestion directa, 2 Publica gestion privada, 3 Privada"),
 "D_GESTION": ("Descripcion del tipo de gestion", "Categorica", "Nominal", "Descripcion de codigo", "Base de la variable derivada prop_publica"),
 "GES_DEP":   ("Codigo de dependencia administrativa", "Categorica", "Nominal", "Codigo", "10 codigos: A1-A4 publicos, B1-B6 privados"),
 "D_GES_DEP": ("Descripcion de la dependencia administrativa", "Categorica", "Nominal", "Descripcion de codigo", "Sector Educacion, Particular, Municipalidad, Comunidad, ..."),
 "DIRECTOR":  ("Nombre del director del servicio", "Categorica", "Nominal", "Texto", "35.62% vacio; dato personal, se excluye del modelo"),
 "TELEFONO":  ("Telefono de contacto", "Categorica", "Nominal", "Texto", "79.12% vacio; se descarta"),
 "EMAIL":     ("Correo electronico de contacto", "Categorica", "Nominal", "Texto", "94.55% vacio; se descarta"),
 "PAGWEB":    ("Pagina web del servicio", "Categorica", "Nominal", "Texto", "98.74% vacio; se descarta"),
 "DIR_CEN":   ("Direccion del local del servicio", "Categorica", "Nominal", "Texto", "Texto libre, no normalizado"),
 "LOCALIDAD": ("Localidad declarada", "Categorica", "Nominal", "Texto", "53.71% vacio"),
 "CODCP_INEI":("Codigo INEI del centro poblado", "Categorica", "Nominal", "Identificador", "12.48% vacio; llave hacia el directorio de CCPP del INEI"),
 "CODCCPP":   ("Codigo interno del centro poblado", "Categorica", "Nominal", "Identificador", "0.72% vacio"),
 "CEN_POB":   ("Nombre del centro poblado", "Categorica", "Nominal", "Texto", "30,510 valores; no normalizado"),
 "AREA_CENSO":("Codigo de area geografica", "Categorica", "Nominal binaria", "Binaria", "1 Urbana, 2 Rural"),
 "DAREACENSO":("Descripcion del area geografica", "Categorica", "Nominal binaria", "Binaria", "Base de la variable derivada prop_rural"),
 "CODGEO":    ("UBIGEO del distrito (2 dpto + 2 prov + 2 dist)", "Categorica", "Nominal", "Identificador", "LLAVE DE AGREGACION DEL MODELO. 1,892 distritos, siempre 6 digitos"),
 "D_DPTO":    ("Nombre del departamento", "Categorica", "Nominal", "Descripcion de codigo", "25 valores"),
 "D_PROV":    ("Nombre de la provincia", "Categorica", "Nominal", "Descripcion de codigo", "196 valores"),
 "D_DIST":    ("Nombre del distrito", "Categorica", "Nominal", "Descripcion de codigo", "1,736 nombres para 1,892 ubigeos: HAY HOMONIMOS, no unir por nombre"),
 "D_REGION":  ("Direccion Regional de Educacion", "Categorica", "Nominal", "Descripcion de codigo", "26 valores"),
 "CODOOII":   ("Codigo de la UGEL de pertenencia", "Categorica", "Nominal", "Identificador", "248 valores; corte administrativo alternativo"),
 "D_DREUGEL": ("Nombre de la DRE o UGEL", "Categorica", "Nominal", "Descripcion de codigo", "247 valores"),
 "NLAT_IE":   ("Latitud del servicio educativo", "Numerica", "Razon", "Continua", "1,302 registros en 0.000000 (nulo codificado); todos inactivos"),
 "NLONG_IE":  ("Longitud del servicio educativo", "Numerica", "Razon", "Continua", "1,302 registros en 0.000000 (nulo codificado); todos inactivos"),
 "TIPOPROG":  ("Codigo del tipo de programa no escolarizado", "Categorica", "Nominal", "Codigo", "69.12% vacio POR DISENO: solo aplica a NIV_MOD = A5"),
 "D_TIPOPROG":("Descripcion del tipo de programa no escolarizado", "Categorica", "Nominal", "Descripcion de codigo", "PRONOEI, SET, Ciclo I/II, PIETBAF, Ludoteca..."),
 "COD_TUR":   ("Codigo de turno de atencion", "Categorica", "Nominal", "Codigo", "11 Manana, 12 Tarde, 13 Manana-Tarde, 14 Noche, 15/16/17 combinaciones"),
 "D_COD_TUR": ("Descripcion del turno", "Categorica", "Nominal", "Descripcion de codigo", "87.5% Manana"),
 "NRORUC":    ("RUC del promotor (gestion privada)", "Categorica", "Nominal", "Identificador", "95.04% vacio POR DISENO: solo gestion privada"),
 "RZSOCIAL":  ("Razon social del promotor", "Categorica", "Nominal", "Texto", "94.09% vacio POR DISENO"),
 "PROMOTOR":  ("Nombre del promotor", "Categorica", "Nominal", "Texto", "89.84% vacio POR DISENO"),
 "ESTADO":    ("Codigo de situacion del servicio", "Categorica", "Nominal binaria", "Binaria", "1 Activo, 2 Inactivo"),
 "D_ESTADO":  ("Descripcion de la situacion del servicio", "Categorica", "Nominal binaria", "Binaria", "FILTRO DE INCLUSION: se trabaja con Activo (115,453)"),
 "FECHAREG":  ("Fecha de registro del servicio", "Categorica", "Intervalo (fecha como texto)", "Fecha", "12.09% vacio; contiene basura ('dd-mm-2007') y fechas imposibles (01-01-1899)"),
 "FECHA_ACT": ("Fecha de actualizacion del padron", "Categorica", "Intervalo (fecha como texto)", "Fecha", "CONSTANTE 05-08-2026: varianza cero, se descarta"),
}

# --- 2. F2: metadatos por bloque de la Cedula 1A ---------------------------
BLOQUES_F2 = {
 "A": "Bloque A - Funcionamiento del servicio, articulacion con salud y programas sociales, continuidad",
 "B": "Bloque B - Gestion institucional, participacion de las familias y convivencia escolar",
 "C": "Bloque C - Practica pedagogica, planificacion, acompanamiento y formacion docente",
}

IDENT_F2 = {
 "COD_MOD":   ("Codigo modular del servicio censado", "Identificador"),
 "ANEXO":     ("Numero de anexo", "Identificador"),
 "CODLOCAL":  ("Codigo del local escolar", "Identificador"),
 "NROCED":    ("Numero de cedula del Censo Escolar", "Identificador"),
 "CEN_EDU":   ("Nombre del servicio educativo", "Texto"),
 "DISTRITO":  ("Nombre del distrito declarado", "Texto"),
 "LOCALIDAD": ("Localidad declarada", "Texto"),
 "ALUM_35IIC":("Preliminar: atiende alumnos de 3 a 5 anos en IIEE de ciclo I", "Binaria"),
 "ALUM_02IC": ("Preliminar: atiende alumnos de 0 a 2 anos en ciclo I", "Binaria"),
 "PRO_CONVER":("Preliminar: el programa fue convertido a IIEE", "Binaria"),
 "ANIOCONVER":("Anio de conversion del programa a IIEE", "Fecha"),
 "NIV_MOD":   ("Codigo de nivel y modalidad", "Codigo"),
 "GES_DEP":   ("Codigo de dependencia administrativa", "Codigo"),
 "AREA_CENSO":("Codigo de area geografica", "Binaria"),
 "CODGEO":    ("UBIGEO del distrito", "Identificador"),
 "CODOOII":   ("Codigo de la UGEL", "Identificador"),
}

def clasificar_f2(col, serie):
    """Devuelve (descripcion, tipo_general, escala, rol_o_subtipo, observaciones)
    para una columna de la Cedula 1A, a partir de su nombre y de sus valores."""
    llenos  = serie[serie != ""]
    valores = set(llenos.unique())
    m = re.match(r"^P(\d{3})([ABC])(.*)$", col)
    if m is None:
        d, rol = IDENT_F2.get(col, ("Columna de identificacion", "Identificador"))
        return (d, "Categorica", "Nominal", rol, "Columna de identificacion o clasificacion")

    num, bloque, sufijo = m.group(1), m.group(2), m.group(3)
    preg = "P%s%s" % (num, bloque)
    ctx = "%s | pregunta %s de la cedula 1A" % (BLOQUES_F2[bloque], preg)

    # 1) Texto libre: se reconoce por la longitud de las respuestas, no por el sufijo
    #    (los sufijos reales son _OT, _5O, _S_6O, _10_O, _SI_A, _NO_O ... demasiado variados).
    if len(llenos) and llenos.str.len().max() > 20:
        return ("Respuesta abierta 'Otro, especifique' de %s" % preg,
                "Categorica", "Nominal", "Texto libre",
                ctx + ". Vacio = no eligio 'Otro'. Requiere normalizacion antes de usarse")

    if valores and valores <= {"SI", "NO", "NS"}:
        cond = bool(sufijo)
        return ("Respuesta a %s%s" % (preg, " (subpregunta %s)" % sufijo if cond else ""),
                "Categorica", "Nominal binaria", "Binaria SI/NO",
                ctx + (". Condicionada: vacio = no aplica (salto)" if cond
                       else ". Vacio = no respuesta real"))

    # 2) Casilla de marca: la proporcion de 'X' se mide SOBRE LOS VALORES LLENOS,
    #    porque muchas alternativas se marcan en menos del 1% de los servicios.
    if (valores and valores <= {"X", "0", "1", ".", "2"}
            and len(llenos) and (llenos == "X").mean() > 0.5):
        return ("Alternativa %s de la pregunta %s" % (sufijo.lstrip("_"), preg),
                "Categorica", "Nominal binaria", "Casilla de marca",
                ctx + ". X = marcada, VACIO = NO MARCADA (equivale a 0), NO es dato faltante. "
                      "Contiene valores contaminados: normalizar con (valor == 'X')")

    if valores and all(v.isdigit() for v in valores):
        return ("Cantidad o codigo reportado en %s%s" % (preg, sufijo),
                "Numerica" if len(valores) > 6 else "Categorica",
                "Razon" if len(valores) > 6 else "Nominal",
                "Conteo" if len(valores) > 6 else "Codigo",
                ctx + ". Almacenado como texto: convertir con pd.to_numeric(errors='coerce'). "
                      "Vacio = pregunta no aplicaba")

    return ("Respuesta a %s%s" % (preg, sufijo), "Categorica", "Nominal", "Por verificar",
            ctx + ". Patron de respuesta no estandar: revisar contra la cedula oficial")

# --- 3. Construccion del diccionario --------------------------------------
def construir_diccionario(datos, nombre_fuente, curado=None, clasificador=None):
    filas = []
    for col in datos.columns:
        serie = datos[col].astype(str).str.strip().replace({"None": "", "nan": ""})
        pct = round((serie == "").mean() * 100, 2)
        ejemplos = ", ".join(map(str, serie[serie != ""].value_counts().head(3).index.tolist()))[:80]

        if curado and col in curado:
            desc, tipo, escala, rol, obs = curado[col]
        elif clasificador is not None:
            desc, tipo, escala, rol, obs = clasificador(col, serie)
        else:
            desc, tipo, escala, rol, obs = ("", "", "", "", "")

        filas.append({
            "fuente": nombre_fuente,
            "variable": col,
            "descripcion": desc,
            "tipo_general": tipo,
            "escala": escala,
            "rol_o_subtipo": rol,
            "tipo_python": str(datos[col].dtype),
            "porcentaje_vacio": pct,
            "valores_distintos": int(serie[serie != ""].nunique()),
            "valores_frecuentes": ejemplos,
            "observaciones": obs,
        })
    return pd.DataFrame(filas)

dicc_f1 = construir_diccionario(df,  "F1_Padron_web",  curado=DESC_F1)
dicc_f2 = construir_diccionario(df1, "F2_Lineal_1A_1", clasificador=clasificar_f2)

diccionario_datos = pd.concat([dicc_f1, dicc_f2], ignore_index=True)

print("Diccionario de datos:", diccionario_datos.shape[0], "variables documentadas")
print("\nDistribucion de roles en F2 (Cedula 1A):")
display(dicc_f2["rol_o_subtipo"].value_counts())

print("\nF1 - Padron_web (45 variables):")
display(dicc_f1)

print("\nF2 - Lineal_1A_1 (muestra de cada tipo de patron):")
display(dicc_f2.groupby("rol_o_subtipo").head(2).sort_values("rol_o_subtipo"))
```

## 1. Problema

El sistema educativo peruano distribuye sus servicios de forma muy desigual en el territorio. La oferta se concentra en zonas urbanas y capitales de provincia, mientras que amplias zonas rurales dependen de servicios no escolarizados o de instituciones con condiciones mínimas de funcionamiento. Esta desigualdad se discute normalmente en términos agregados (departamento o región) y no a nivel **distrital**, que es la unidad donde efectivamente se ejecuta la inversión pública y donde una familia decide si puede o no matricular a su hijo.

El proyecto se concentra en **Educación Inicial**, por tres razones:

1. Es el nivel donde la brecha es más severa: de los **53,650** servicios de inicial activos en el país, **17,523 (32.7%)** no son escolarizados (PRONOEI, SET, entornos comunitarios y familiares), es decir, no son un jardín con local e infraestructura propia.
2. Es el nivel más rural del sistema: **55.6%** de los servicios de inicial están en área rural, y en departamentos como Huancavelica (**93.7%**), Amazonas (**92.1%**) y Cajamarca (**87.3%**) la oferta es casi enteramente rural, frente a Callao (**0%**) y Lima (**8.6%**).
3. Es el único nivel para el que disponemos de una segunda fuente con información de condiciones de funcionamiento del servicio (la Cédula 1A del Censo Escolar).

**Pregunta de investigación:** ¿existen agrupamientos (*clusters*) de distritos con perfiles similares de oferta de Educación Inicial, y qué distritos concentran las mayores brechas según densidad de servicios, grado de escolarización, tipo de gestión, ruralidad y condiciones de funcionamiento del servicio?

**Objetivo general:** identificar y caracterizar perfiles distritales de oferta de Educación Inicial para detectar los distritos con déficit estructural.

**Objetivos específicos:**

- Construir indicadores distritales a partir del padrón: número de servicios de inicial activos, proporción no escolarizada, proporción rural, proporción pública, proporción con local escolar registrado.
- Añadir, para los distritos cubiertos por el Censo Escolar, indicadores de condiciones del servicio derivados de la Cédula 1A (materiales educativos, articulación con salud y programas sociales, participación de las familias, acompañamiento pedagógico, casos reportados de violencia).
- Segmentar los **1,886** distritos con oferta de inicial escolarizada activa mediante *clustering* no supervisado.
- Caracterizar los clusters y señalar cuáles corresponden a perfiles de déficit.

**Justificación:** el resultado permite priorizar dónde convertir programas no escolarizados en jardines, dónde reforzar el acompañamiento y dónde focalizar programas de acceso. La magnitud del problema se ve en un dato del propio padrón: la **mediana** de servicios de inicial escolarizada por distrito es de **9**, pero **312 distritos** (16.5%) tienen **2 o menos**, y **6 distritos** no tienen **ninguno**, mientras el máximo llega a **601** en un solo distrito de Lima.

> **Cambio de alcance respecto de la versión anterior de este documento.** El diseño original planteaba calcular ratios alumno/docente y alumno/sección a nivel nacional. El perfilado mostró que **ninguna de las fuentes disponibles contiene matrícula, docentes ni secciones**: `Padron_web.dbf` es un directorio administrativo (45 campos, ninguno de conteo) y `Lineal_1A_1.dbf` es un cuestionario de gestión, no un padrón de matrícula. Se reencuadró el problema a lo que las fuentes sí permiten medir. Los ratios quedan como línea de trabajo futura, condicionada a incorporar el módulo de matrícula del Censo Escolar.

## 7. Problemas encontrados

### 7.1 Estructura e identificadores
- **Duplicados aparentes:** un mismo `COD_MOD` aparece varias veces con distinto `ANEXO`. Agrupar solo por COD_MOD subcuenta servicios.
- **Ceros a la izquierda destruidos** al abrir el CSV en Excel o al leer sin `dtype=str`: `COD_MOD`, `CODLOCAL` y `UBIGEO` se corrompen. Es el error más costoso porque rompe los joins de forma silenciosa.
- **Relación 1:N** entre local escolar y servicios: unir F1 con F3 sin cuidado infla los conteos.

### 7.2 Valores faltantes
- Coordenadas (`NLAT_IE`, `NLONG_IE`) ausentes o en `(0,0)`, típicamente en zonas rurales — justo donde está la brecha, lo que genera **sesgo de selección** si se descartan esas filas.
- `TDOCENTE`, `TALUMNO`, `TSECCION` vacíos en servicios no escolarizados (PRONOEI) y en parte de la gestión privada, que reporta con menor tasa de respuesta.
- **Ambigüedad entre `0` real y nulo codificado como 0:** un colegio con 0 docentes es probablemente dato faltante, no un colegio sin docentes.

### 7.3 Tipos mal detectados
- `UBIGEO` leído como `int64` (Amazonas `010101` → `10101`).
- Conteos numéricos leídos como `object` por separadores de miles o espacios.
- Fechas de registro como texto en formatos mixtos.
- **Encoding Latin-1 no declarado:** `Áncash` → `A?ncash`, lo que rompe cualquier join por nombre de distrito.

### 7.4 Categorías raras o inconsistentes
- Categorías de baja frecuencia en `D_NIV_MOD` (EBE, ETP, Superior Pedagógica) no comparables con EBR; conviene separarlas o excluirlas del cálculo de ratios.
- Variantes ortográficas y de acentuación en nombres de distrito (`Nasca`/`Nazca`, con y sin tildes) — razón adicional para unir por UBIGEO y **nunca** por nombre.
- `D_ESTADO` con múltiples valores fuera de "Activo": hay que definir explícitamente el criterio de inclusión.
- Distritos de creación reciente presentes en el padrón MINEDU pero ausentes en las proyecciones INEI, o viceversa → filas huérfanas en el join.

### 7.5 Inconsistencias lógicas y valores extremos
- `ratio_alum_doc` infinito o indefinido cuando `TDOCENTE = 0`.
- Ratios extremos (> 60 alumnos/docente) que pueden ser reales en zonas críticas o errores de captura: revisarlos manualmente antes de winsorizar, porque son precisamente los casos de interés.
- Servicios cuyo `UBIGEO` no coincide con el departamento declarado en `D_DPTO`.
- Distritos con muy pocos servicios, donde los promedios son inestables.

### 7.6 Sesgos y limitaciones del diseño
- El padrón mide **oferta registrada**, no acceso efectivo: un distrito con muchas IIEE puede seguir teniendo baja asistencia.
- Sin la fuente de población (F3), la densidad de IIEE no es un indicador de brecha sino de **tamaño del distrito**.
- No hay variable de distancia ni tiempo de traslado, que es el determinante real del acceso en zonas rurales andinas y amazónicas.
- Las variables numéricas están en escalas muy distintas → **estandarización obligatoria** antes de K-means; evaluar K-prototypes si se mantienen variables categóricas.

## 8. Decisiones de limpieza justificadas

Se documentan las decisiones tomadas sobre las tres fuentes, priorizando F4 (denuncias escolares)

### 8.1 Qué se corrigió
- **Formato de texto de `UGEL`** (F4): se normalizó a un único patrón (`"NN Nombre"`, mayúsculas, sin tildes, sin doble espacio), eliminando el guion en las 10 UGEL que lo traían, para que el campo sea comparable consigo mismo y con otras fuentes.
- **Prefijo `DRE`/`GRE`** (F1 vs F4): se normalizó `GRE` → `DRE` en `D_DREUGEL` del padrón antes de comparar contra `denuncias_escolares`, porque ambos prefijos designan la misma entidad administrativa (Dirección/Gerencia Regional de Educación); mantenerlos distintos rompería el join en 9 regiones.
- **Acentos y mayúsculas** en `DRE`/`UGEL` (F4) y `D_DREUGEL` (F1): se normalizaron a un mismo estándar (mayúsculas, sin tildes) exclusivamente para efectos de *matching*; se conserva la columna original sin tocar para reporte y visualización.

### 8.2 Qué se imputó
- **No se imputó ningún valor en F4.** No hay celdas vacías (0% de faltantes en las 8 columnas), así que no hay nada que imputar.
- En F1/F2, la imputación sigue limitada a los casos ya descritos en la sección 7.2 (coordenadas en `(0,0)`, docentes/secciones vacíos en no escolarizados): **se decidió no imputar coordenadas ni conteos**, y en su lugar se dejará un *flag* booleano (`coord_valida`, `dato_reportado`) porque imputar la ubicación o el conteo de un servicio inventaría información que no existe en la fuente.

### 8.3 Qué se eliminó
- **No se eliminaron las 7,282 filas duplicadas exactas de F4.** Al no existir un identificador de caso (`ID_CASO`), eliminar duplicados asumiría —sin evidencia— que son el mismo caso repetido, cuando pueden ser casos distintos con el mismo perfil categórico (misma fecha, misma UGEL, mismo tipo de violencia). Eliminar reportes reales de violencia escolar es el error más costoso posible en esta base: se prefiere sobreconteo documentado a subconteo silencioso.
- Si el objetivo del análisis exige un conteo conservador (por ejemplo, para comparar tasas entre UGEL), se calculará también un conteo alternativo con duplicados exactos colapsados a 1, y **se reportarán ambos números lado a lado**, nunca solo uno.
- Se mantiene la decisión ya tomada en F1: no eliminar los 65,339 registros con `D_ESTADO != 'Activo'`; se excluyen del cálculo de indicadores pero se conservan en el dataframe base.

### 8.4 Qué se mantuvo y por qué
- Se mantuvo el registro `UGEL == "DRE Callao"` (1,328 filas) tal como viene: no es un error, Callao no se subdivide en UGEL y la propia DRE cumple esa función administrativa, lo cual es coherente con que `D_DREUGEL` en el padrón también use `"DRE CALLAO"` para esos colegios.
- Se mantuvo la categoría `"Con uso de armas"` clasificada tanto en `Física` como `Psicológica`: no se fuerza a una sola categoría porque la ambigüedad refleja un criterio real del formulario SíseVe (amenaza con arma sin contacto físico vs. agresión con arma), no un error de captura.
- Se mantuvieron las 4 UGEL de F4 que no encuentran par exacto en F1 tras la normalización (`UGEL 03 Cercado`, `UGEL Antonio Raimondi`, `UGEL Carlos F. Fitzcarrald`, `UGEL Pichari‑Kimbiri‑Villa Virgen`) sin forzar un match aproximado automático: se documentan como pendientes de un diccionario de equivalencias manual (ver sección 9), porque un *fuzzy match* automático sobre nombres administrativos puede unir UGEL distintas silenciosamente.


```python

# 8. IMPLEMENTACIÓN DE LAS DECISIONES DE LIMPIEZA

import unicodedata

def normaliza_texto(s):
    """Mayusculas, sin tildes, sin guion, espacios simples. Solo para matching."""
    s = str(s).upper().strip()
    s = s.replace("-", " ")
    s = " ".join(s.split())
    s = "".join(c for c in unicodedata.normalize("NFD", s) if unicodedata.category(c) != "Mn")
    return s

# Copia de trabajo: se conservan DRE/UGEL originales, se agregan versiones normalizadas
df2_limpio = df2.copy()
df2_limpio["DRE_norm"] = df2_limpio["DRE"].apply(normaliza_texto)
df2_limpio["UGEL_norm"] = df2_limpio["UGEL"].apply(normaliza_texto)

# Se flaggean duplicados exactos SIN eliminarlos (decision documentada en 8.3)
df2_limpio["es_duplicado_exacto"] = df2_limpio.duplicated(keep=False)
print("Duplicados marcados (no eliminados):", df2_limpio["es_duplicado_exacto"].sum())

# Conteo alternativo (conservador), reportado aparte, nunca reemplazando al original
n_casos_reportados = len(df2_limpio)
n_casos_sin_duplicado_exacto = len(df2_limpio.drop_duplicates())
print(f"Conteo de casos, versión completa: {n_casos_reportados}")
print(f"Conteo de casos, colapsando duplicados exactos: {n_casos_sin_duplicado_exacto}")

display(df2_limpio.head())
```

## 9. Base integrada inicial: incorporación de F4 (denuncias escolares)

F4 tiene una **granularidad distinta** a F1/F2 (caso reportado vs. servicio educativo) y **no comparte ninguna llave exacta tipo `COD_MOD`**: solo trae texto libre de DRE/UGEL. Por eso no puede integrarse fila a fila como F1↔F2; el plan es una **integración por agregación** a nivel UGEL.

**Plan de integración:**
1. Normalizar `D_DREUGEL` (F1) y `DRE`+`UGEL` (F4) con la misma función (`normaliza_texto`): mayúsculas, sin tildes, sin guiones, prefijo `GRE`→`DRE`.
2. Agregar F4: número de casos reportados por `UGEL_norm` (y opcionalmente por `TIPO_VIOLENCIA`).
3. Agregar F1: número de servicios activos por `D_DREUGEL_norm` (mismo campo usado como llave de UGEL en el padrón).
4. Unir ambas tablas agregadas por la llave de texto normalizada (`UGEL_norm`), **no** al nivel de `COD_MOD` porque F4 no lo trae.
5. Reportar explícitamente el **porcentaje de UGEL y de filas que no encuentran par**, en vez de descartarlas en silencio.

Se probó el paso 1–4 para medir qué tan viable es la llave de texto:


```python

# 9 PRUEBA DE INTEGRACIÓN F1 <-> F4 POR UGEL NORMALIZADA

# UGEL normalizadas presentes en cada fuente
ugel_f4 = set(df2_limpio["UGEL_norm"].unique())

def normaliza_dreugel_f1(s):
    s = normaliza_texto(s)
    s = s.replace("GRE ", "DRE ", 1) if s.startswith("GRE ") else s
    return s

ugel_f1 = set(normaliza_dreugel_f1(rec["D_DREUGEL"]) for rec in tabla_f1)

interseccion = ugel_f4 & ugel_f1
sin_match = ugel_f4 - ugel_f1

print(f"UGEL distintas en F4: {len(ugel_f4)}")
print(f"UGEL distintas en F1 (D_DREUGEL normalizado): {len(ugel_f1)}")
print(f"UGEL de F4 que encuentran par en F1: {len(interseccion)} ({len(interseccion)/len(ugel_f4)*100:.1f}%)")
print(f"UGEL de F4 SIN par en F1: {sorted(sin_match)}")

filas_sin_match = (~df2_limpio["UGEL_norm"].isin(ugel_f1)).sum()
print(f"\nFilas de F4 cubiertas por el join: {len(df2_limpio) - filas_sin_match} "
      f"({(len(df2_limpio)-filas_sin_match)/len(df2_limpio)*100:.1f}%)")
print(f"Filas de F4 SIN cobertura (quedan fuera del análisis integrado): {filas_sin_match} "
      f"({filas_sin_match/len(df2_limpio)*100:.1f}%)")

# --- Base integrada inicial a nivel UGEL
casos_por_ugel = (
    df2_limpio.groupby("UGEL_norm")
    .size()
    .rename("n_casos_violencia")
    .reset_index()
)

servicios_f1 = pd.DataFrame(iter(tabla_f1))
servicios_f1["UGEL_norm"] = servicios_f1["D_DREUGEL"].apply(normaliza_dreugel_f1)
servicios_activos_por_ugel = (
    servicios_f1[servicios_f1["D_ESTADO"] == "Activo"]
    .groupby("UGEL_norm")
    .size()
    .rename("n_servicios_activos")
    .reset_index()
)

base_integrada_ugel = pd.merge(
    servicios_activos_por_ugel, casos_por_ugel, on="UGEL_norm", how="outer", indicator=True
)
print("\nResultado del merge (base integrada a nivel UGEL):")
display(base_integrada_ugel["_merge"].value_counts())
display(base_integrada_ugel.head())
```
